```python
import numpy as np
import pandas as pd
```

load kiya dataset

```python
train_df=pd.read_csv('train.csv')
```

```python
train_df.isnull().sum()
```

binary encoding

```python

# Binary Encoding X

train_df['Home_Charging_Possible'] = train_df['Home_Charging_Possible'].map({
    'Yes': 1,
    'No': 0
})

train_df['Subsidy_Available'] = train_df['Subsidy_Available'].map({
    'Yes': 1,
    'No': 0
})
```

```python
train_df['Will_Buy_EV'] = train_df['Will_Buy_EV'].map({
    'Yes': 1,
    'No': 0
})
```

split kiya

```python
X = train_df.iloc[:, :-1].values
Y = train_df.iloc[:, -1].values
```

```python
print(X[0])
print(Y[0])
```

train test split

```python
from sklearn import preprocessing
from sklearn.model_selection import train_test_split
X_train, X_val, y_train, y_val = train_test_split(X, Y, random_state=0,test_size=0.2,stratify=Y)
```

one hot encoding

```python
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder

ct = ColumnTransformer(
    transformers=[
        ('encoder', OneHotEncoder(), [8, 9, 10, 13])
    ],
    remainder='passthrough'
)

X_train = np.array(ct.fit_transform(X_train))
```

```python
X_val = np.array(ct.transform(X_val))
```
